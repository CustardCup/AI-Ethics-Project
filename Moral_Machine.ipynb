###1. The Experiment

The moral machine is a classic thought experiment from the field of artificial intelligence and automation. It sheds light on how human decision making and values are inseparable from autonomous technologies. Here is the idea:

*Imagine that you’re driving to work in your 2020 Prius, it’s not autonomous and you’re in complete control. You drive like a law abiding citizen but as you head down the last few roads to your company, a tree falls shortly in front of you and blocks the road. Out of panic, you swerve and kill a pedestrian.*



JPEG image-463B-A0CD-BD-0.jpeg

Ask yourself a moral question here: who is to blame in the death of the pedestrian? Is it an accident? Did you even have time to see or react to the pedestrian? For the sake of this thought experiment, I want you to imagine that this situation is a tragic incident where you made a subconscious decision to save your own life resulting in the death of another. Let’s re-do this experiment.

*Imagine that you’re driving to work in a 2050 Prius, it’s autonomous and it’s not even your car, it’s a government ride service that you take to and from work. It drives better than you, more safely than you, but as you head down the last few roads to your company, a tree falls shortly in front of you and blocks the road. The car knows that it is presented with a fatal dilemma, it can think billions of times faster than you and has profiled both you and the pedestrian. It weighs its options and has makes a moral judgement, taking immediate action before the tree even hits the ground.*

```python
# @markdown Import Simulation Data (run this)
import json
!git clone https://github.com/CustardCup/AI-Ethics-Project
%cd AI-Ethics-Project
with open('moral-machine.json') as f:
    data = json.load(f)
```

```python
from collections import Counter

tally = {
    "charType": Counter(),
    "ageGroup": Counter(),
    "profession": Counter(),
    "sex": Counter(),
    "pregnant": Counter(),
}

def add_survivors(survivors):
    for survivor in survivors:
        for trait in tally:
            tally[trait][survivor[trait]] += 1
```

```python
# Modified Function
def decision(passengers, pedestrians):
    if len(passengers) > len(pedestrians):
        survivors = passengers

    #Selects to save most children in a tie on numbers

    elif len(passengers) == len(pedestrians):
        child_tally_pass = 0
        child_tally_pede = 0
        for person in passengers:
            if person["ageGroup"] == "child":
                child_tally_pass += 1
        for person in pedestrians:
            if person["ageGroup"] == "child":
                child_tally_pede += 1
        if child_tally_pass > child_tally_pede:
            survivors = passengers
        else:
            survivors = pedestrians

    else:
        survivors = pedestrians
    return survivors

# Tally Function
def decision_tally(survivors):
    add_survivors(survivors)
```

```python
tally = {
    "charType": Counter(),
    "ageGroup": Counter(),
    "profession": Counter(),
    "sex": Counter(),
    "pregnant": Counter(),
}

for scenario in data.values():
  decision_tally(decision(scenario["passengers"], scenario["pedestrians"]))

print(tally)
```
